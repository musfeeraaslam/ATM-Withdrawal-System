```python
def withdraw(amount,balance):
  if amount <=0:
    print("Error--Amount should be greater than zero")
    return balance
  if amount > balance:
    print("Error-- Insufficent balance")
    return balance

  balance = balance - amount
  print("Withdrawl successfully")
  print(f"Remaining balance is: {balance}")

  if balance <500:
    print("Warning: Low Balance")
```

```python
balance= float(input("Enter the total balance: "))
amount= float(input("Enter wihdrawal amount: "))

withdraw(amount,balance)
```
